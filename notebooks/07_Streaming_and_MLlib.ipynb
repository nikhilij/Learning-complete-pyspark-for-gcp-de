## Structured Streaming overview

Structured Streaming lets Spark process streaming data using the same DataFrame APIs as batch jobs.


```python
from pyspark.sql import SparkSession
from pyspark.sql import functions as F

spark = SparkSession.builder.appName('StreamingDemo').master('local[*]').getOrCreate()

schema = 'timestamp STRING, user_id INT, event STRING'
stream_df = spark.readStream.schema(schema).csv('data/stream_input')

transformed = stream_df.groupBy('event').count()
query = transformed.writeStream.outputMode('complete').format('console').start()
query.stop()

spark.stop()
```

## MLlib example

MLlib supports distributed machine learning and feature engineering.


```python
from pyspark.sql import SparkSession
from pyspark.ml.feature import VectorAssembler
from pyspark.ml.regression import LinearRegression

spark = SparkSession.builder.appName('MLlibDemo').master('local[*]').getOrCreate()

data = [
    (1.0, 1.0, 10.0),
    (2.0, 2.0, 20.0),
    (3.0, 3.0, 30.0),
    (4.0, 4.0, 40.0),
]

df = spark.createDataFrame(data, ['x', 'y', 'label'])
assembler = VectorAssembler(inputCols=['x', 'y'], outputCol='features')
training_df = assembler.transform(df)
training_df = training_df.select('features', 'label')

lr = LinearRegression(featuresCol='features', labelCol='label', maxIter=10, regParam=0.0)
model = lr.fit(training_df)
print('Coefficients:', model.coefficients)
print('Intercept:', model.intercept)

spark.stop()
```
