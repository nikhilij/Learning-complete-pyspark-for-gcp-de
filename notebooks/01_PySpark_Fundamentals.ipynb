## Spark architecture

Spark uses a master/worker model. The main components are:

- Driver program: the process that runs your main application logic
- Executors: worker processes that perform tasks
- Cluster manager: allocates resources (YARN, Kubernetes, standalone, local)
- RDD/DataFrame: distributed data abstractions
- DAG scheduler: plans work and optimizes execution


```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName('PySpark-Fundamentals') \
    .master('local[*]') \
    .config('spark.sql.shuffle.partitions', '4') \
    .getOrCreate()

print('Spark version:', spark.version)
spark.stop()
```

## Lazy evaluation

Spark does not execute transformations immediately. Instead, it builds a logical plan and waits until an action triggers execution.


```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName('LazyEval').master('local[*]').getOrCreate()

data = [1, 2, 3, 4, 5]
rdd = spark.sparkContext.parallelize(data)

mapped = rdd.map(lambda x: x * 2)
filtered = mapped.filter(lambda x: x > 5)

print('No execution yet; only a plan has been built.')
print(filtered.collect())

spark.stop()
```

## Transformations vs actions

### Transformations
- `map`
- `filter`
- `select`
- `withColumn`
- `groupBy`

These build a new dataset without executing immediately.

### Actions
- `collect`
- `show`
- `count`
- `take`
- `save`

These trigger execution.


```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName('TransformActionDemo').master('local[*]').getOrCreate()

df = spark.createDataFrame([(1, 'Alice'), (2, 'Bob'), (3, 'Charlie')], ['id', 'name'])
filtered_df = df.filter(df.id > 1)
print('Filtered rows:')
filtered_df.show()
print('Row count:', filtered_df.count())

spark.stop()
```

## Why Spark is efficient

Spark builds a query plan, optimizes it, and distributes work across partitions. This makes it efficient for large-scale analytics and ETL jobs.
