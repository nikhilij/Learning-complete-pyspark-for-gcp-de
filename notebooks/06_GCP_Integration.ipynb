# 06. GCP Integration: GCS, BigQuery, and ETL patterns

Modern data engineering work frequently integrates Spark with cloud storage and warehouse services. This notebook introduces the common GCP patterns.

## GCP focus areas
- Cloud Storage (GCS) for raw and processed data
- BigQuery for analytics and warehouse workflows
- Dataproc or serverless Spark for job execution
- IAM and credentials management


## Reading and writing to local Parquet

This pattern mirrors how Spark reads and writes files in production.


```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName('GCSIntegration').master('local[*]').getOrCreate()

df = spark.createDataFrame([(1, 'Alice', 2500), (2, 'Bob', 3200)], ['id', 'name', 'salary'])
df.write.mode('overwrite').parquet('data/output/employees.parquet')
read_df = spark.read.parquet('data/output/employees.parquet')
read_df.show()

spark.stop()
```

## Reading CSV files


```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName('CSVReader').master('local[*]').getOrCreate()

df = spark.read.csv('data/sample_data.csv', header=True, inferSchema=True)
df.show(5)

spark.stop()
```

## BigQuery pattern

In production, Spark can read from and write to BigQuery using the Spark BigQuery connector. Typical setup includes credentials, project configuration, and proper table schema management.


```python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName('BigQueryPattern') \
    .master('local[*]') \
    .getOrCreate()

df = spark.createDataFrame([(1, 'A'), (2, 'B')], ['id', 'name'])
print('Example BigQuery-ready dataframe created successfully.')
print(df.collect())

spark.stop()
```

## Typical GCP ETL flow
1. raw data lands in GCS
2. Spark cleans and transforms the data
3. data is written as Parquet or Delta
4. analytics tables are loaded into BigQuery
5. dashboards and pipelines consume those curated datasets
