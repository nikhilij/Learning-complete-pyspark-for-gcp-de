# 02. RDDs and DataFrames

This notebook introduces the two key distributed data abstractions in Spark: RDDs and DataFrames.

## What is an RDD?
RDDs are low-level distributed collections of objects. They provide strong control but less structure than DataFrames.

## What is a DataFrame?
DataFrames are structured tables with named columns and schema information. They are usually the preferred interface in modern PySpark.


```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName('RDD-DataFrame').master('local[*]').getOrCreate()

data = [1, 2, 3, 4, 5]
rdd = spark.sparkContext.parallelize(data)
print('RDD elements:', rdd.collect())

df = spark.createDataFrame([(1, 'A'), (2, 'B'), (3, 'C')], ['id', 'letter'])
df.show()

spark.stop()
```

## RDD example

RDDs are useful for lower-level transformations and understanding distributed execution.


```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName('RDDExamples').master('local[*]').getOrCreate()

text = ['spark is fast', 'spark is scalable', 'spark is distributed']
rdd = spark.sparkContext.parallelize(text)

words = rdd.flatMap(lambda line: line.split())
word_counts = words.map(lambda w: (w, 1)).reduceByKey(lambda a, b: a + b)

print(word_counts.collect())
spark.stop()
```

## DataFrame example

A DataFrame is more structured and easier to reason about than an RDD. It has named columns and a schema.


```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName('DataFrameBasics').master('local[*]').getOrCreate()

data = [
    (1, 'Alice', 29),
    (2, 'Bob', 34),
    (3, 'Carla', 41)
]

df = spark.createDataFrame(data, ['id', 'name', 'age'])
df.printSchema()
df.show()

spark.stop()
```

## Key differences
- RDDs are low-level and unstructured
- DataFrames have schema and optimization
- DataFrames use Catalyst and Tungsten optimizations
- Most modern PySpark jobs use DataFrames

## Exercise
Create a DataFrame from sales records and sort it by price.


```python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName('SalesDataFrame').master('local[*]').getOrCreate()
sales = [
    (101, 'Laptop', 1200),
    (102, 'Phone', 700),
    (103, 'Tablet', 600),
    (104, 'Monitor', 350)
]

df = spark.createDataFrame(sales, ['order_id', 'product', 'price'])
df.printSchema()
df.orderBy('price', ascending=False).show()

spark.stop()
```
