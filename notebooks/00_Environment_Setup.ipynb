# 00. Environment Setup and First Spark Run

This notebook prepares the local environment for PySpark development and verifies that Spark can start correctly.

## Goals
- confirm Python and Java availability
- create a virtual environment
- install project dependencies
- validate SparkSession startup
- understand local vs cluster execution


````python
import os
import platform
import sys

print('Python version:', sys.version)
print('Platform:', platform.platform())
print('JAVA_HOME:', os.environ.get('JAVA_HOME'))
````

## Create a virtual environment

```bash
python -m venv .venv
source .venv/bin/activate
pip install -r requirements.txt
```

This keeps your notebook environment isolated and reproducible.


````python
from pyspark.sql import SparkSession

spark = SparkSession.builder \
    .appName('PySpark-Environment-Check') \
    .master('local[*]') \
    .config('spark.sql.shuffle.partitions', '4') \
    .getOrCreate()

print('Spark version:', spark.version)
spark.stop()
````

## First Spark program

The SparkSession is the main entry point for DataFrame and SQL operations.


````python
from pyspark.sql import SparkSession

spark = SparkSession.builder.appName('FirstProgram').master('local[*]').getOrCreate()
values = spark.range(1, 6)
print(values.collect())
spark.stop()
````
